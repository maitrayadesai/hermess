# Small-signal analysis

With `small_signal_analysis=True` the operating point is linearized before
the time stepping: the algebraic variables are eliminated, the reduced state
matrix is built, and its eigenvalues, participation factors and a modal
report are attached to the returned model. The analysis lives entirely at
t = 0 s, so a short horizon keeps the run cheap.

The system is the standard IEEE 39-bus benchmark with ten subtransient
machines. The run prints the modal report, sorted by damping ratio, so the
critical electromechanical modes top the list:

```python
import hermess

dae = hermess.simulate(
    "ieee39", T_end=0.5, ts=5e-3, line_dyn=False,
    small_signal_analysis=True, small_signal_figures=False,
)
```

The s-plane shows the same information at a glance. Modes to the right of
the dashed damping cone are the ones to watch, the swing modes between 0.6
and 1.6 Hz:

```python
dae.plot_eigenvalues(damping_ref=0.05)
```

The participation factors say which states carry a mode, i.e. where a
damping problem lives and which machine a controller should act on. Mode 2
of the report is the least damped oscillatory mode; its participation is
shared across machines on both sides of the network, the signature of an
inter-area mode rather than a local one:

```python
dae.participation_table(mode=2)
```

Everything shown here is also available as plain data (`dae.A`,
`dae.eigenvalues`, `dae.state_names`, and `res.small_signal` after
`extract_results`), so a parameter study, a damping-versus-gain sweep for
example, is a loop over `simulate` calls away.
